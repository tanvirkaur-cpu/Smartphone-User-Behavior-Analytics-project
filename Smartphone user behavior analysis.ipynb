```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import os

def login():
    print("You will only get 3 chances")
    user = "tanvirrkr"
    pas = "huihuihui"
    attempts = 3
    logged_in = False

    while attempts > 0:
        username = input("Enter Username: ")
        if username == user:
            password = input("Enter your password: ")
            if password == pas:
                print("Logged in successfully")
                logged_in = True
                break
            else:
                print("Invalid Password")
                attempts -= 1
        else:
            print("Invalid Username")
            attempts -= 1

    return logged_in

if not login():
    print("Too many failed attempts. Exiting.")
    exit()

ans = input("Should we proceed by uploading data? (yes/no): ").strip().lower()
if ans == "yes":
    path = r"C:\Users\Kaurt\OneDrive\Documents\Final Project\Raw Data (User Behaviour) - Form responses 1.csv"
    upload = pd.read_csv(path)
    print("Data uploaded Successfully")
    exit()
else:
    print("Data not uploaded. Try Again")

upload.columns = upload.columns.str.strip()
for col in upload.select_dtypes(['object']):
    upload[col] = upload[col].str.strip()
upload.dropna(inplace=True)
print("Data Uploaded and Cleaned Successfully")

while True:
    print("\n--- Main Menu ---")
    print("1. Demographic Analysis")
    print("2. Screentime Analysis")
    print("3. Behaviour Pattern Analysis")
    print("4. Digital Life Analysis")
    print("5. Health Impact Analysis")
    print("6. Exit")

    try:
        op = int(input("Enter your choice: "))
    except ValueError:
        print("Please enter a valid number.")
        continue

    if op == 1:
        print("\nDEMOGRAPHIC ANALYSIS MENU")
        print("1. Which age group uses phones the most?")
        print("2. Which gender reports higher Screen time?")
        print("3. Which OS is most preferred nowadays?")
        print("4. Back to Main Menu")

        try:
            ch = int(input("Enter your choice: "))
        except ValueError:
            print("Invalid input.")
            continue

        if ch == 1:
            age_counts = upload["Participant Age Group"].value_counts()
            print("\nPhone usage by Age Group:")
            print(age_counts)
            age_counts.plot(kind="bar", color="blue")
            plt.title("Distribution of Users by Age")
            plt.ylabel("Number of users")
            plt.tight_layout()
            plt.show()

        elif ch == 2:
            gendr = upload.groupby("Average Daily Screen-On Time (SOT)")["Gender Identity"].value_counts().unstack()
            print("\nScreen Time Tiers by Gender")
            print(gendr)
            gendr.plot(kind="bar", stacked=True)
            plt.title("Screentime Distribution by Gender")
            plt.tight_layout()
            plt.show()

        elif ch == 3:
            os_data = upload["Primary Mobile Operating System"].value_counts()
            print("\nOS Preference Distribution")
            print(os_data)
            os_data.plot(kind="pie", autopct="%1.2f%%", colors=["pink", "blue","green"])
            plt.title("Most Preferred Operating System")
            plt.tight_layout()
            plt.show()

        elif ch == 4:
            continue

        else:
            print("Invalid choice. Please Try Again")

    elif op == 2:
        print("\nSCREENTIME ANALYSIS MENU")
        print("1. Average Screen Time by age group")
        print("2. High screen time users vs low screen time users")
        print("3. Screen time vs Ergonomic Strain")
        print("4. Back to Main Menu")

        try:
            b = int(input("Enter your Choice: "))
        except ValueError:
            print("Invalid input.")
            continue

        if b == 1:
            actv = upload["Peak Activity Window"].value_counts()
            print("\nPeriod in which device activity is most intensive")
            print(actv)
            actv.plot(kind="pie", autopct="%1.1f%%", colors=["skyblue", "gold", "red", "lightgreen"])
            plt.title("Peak Activity Window")
            plt.tight_layout()
            plt.show()

        elif b == 2:

                    low_mask = upload['Average Daily Screen-On Time (SOT)'].str.contains('<2 hours|2-4 hours', case=False, na=False)
                    high_mask = upload['Average Daily Screen-On Time (SOT)'].str.contains('6-8 hours|8\+ hours', case=False, na=False)
                    
                    low_count = low_mask.sum()
                    high_count = high_mask.sum()
                    
                    comparison_data = pd.Series({
                        'Low Screen time users': low_count,
                        'High Screen time users': high_count
                    })
                    
                    print("\n--- Comparison Results ---")
                    print(comparison_data)
                    
                
                    if comparison_data.sum() > 0:
                        plt.figure(figsize=(8, 5))
                        colors = ['#82ca9d', '#8884d8']
                        bars = plt.bar(comparison_data.index, comparison_data.values, color=colors, edgecolor='black', width=0.6)
                        
                        plt.title('High Screen Time users vs Low Screen Time users', fontweight='bold', fontsize=14)
                        plt.ylabel('Number of Users')
                        plt.grid(axis='y', linestyle='--', alpha=0.7)
                        
                        # Add value labels on top of bars
                        for i in bars:
                            yval = i.get_height()
                            plt.text(i.get_x() + i.get_width()/2, yval + 0.1, int(yval), ha='center', fontweight='bold')
                            
                        plt.tight_layout()
                        plt.show()
                    else:
                        print("\n[!] No data found. Check if 'Average Daily Screen-On Time (SOT)' values match the expected tiers.")
                
        elif b == 3:
               strn = upload.groupby("Physical Ergonomic Strain")["Average Daily Screen-On Time (SOT)"].value_counts().unstack()
               print("\nScreen Time vs Ergonomic Strain")
               print(strn)
               strn.plot(kind="bar", color="skyblue", stacked=True)
               plt.title("SOT vs Strain")
               plt.xlabel("SOT")
               plt.ylabel("Physical Strain")
               plt.tight_layout()
               plt.show()

        elif b == 4:
            continue

        else:
            print("Invalid Choice. Enter a Valid Number")

    elif op == 3:
        print("\nBEHAVIOUR PATTERN ANALYSIS MENU")
        print("1. What is the peak activity window?")
        print("2. Do Users check phone immediately after waking up?")
        print("3. How many people uses their phones during meals?")
        print("4. Back to Main Menu")

        try:
            c = int(input("Enter your choice: "))
        except ValueError:
            print("Invalid input.")
            continue

        if c == 1:
            peak = upload["Peak Activity Window"].value_counts()
            print("\nPeak Activity Window")
            print(peak)
            peak.plot(kind="pie", autopct="%1.1f%%")
            plt.title("Peak Activity Window")
            plt.tight_layout()
            plt.show()

        elif c == 2:
            usrr = upload["Immediate Engagement (Waking)"].value_counts()
            print("\nImmediate Engagement in 15 minutes After waking up")
            print(usrr)
            usrr.plot(kind="bar", color="pink")
            plt.title("Immediate Engagement")
            plt.xlabel("Engagement")
            plt.tight_layout()
            plt.show()

        elif c == 3:
            meal = upload["Meal-Time Device Interaction"].value_counts()
            print("Meal Time Interaction")
            print(meal)
            meal.plot(kind="bar", color="pink")
            plt.title("Meal Time Interaction")
            plt.xlabel("Interaction")
            plt.tight_layout()
            plt.show()

        elif c == 4:
            continue

        else:
            print("Invalid Choice. Please Enter a Valid Number")

    elif op == 4:
        print("\nDIGITAL LIFESTYLE ANALYSIS MENU")
        print("1. Which category accounts for the majority of the usage?")
        print("2. Does people with higher proficiency spend more time online?")
        print("3. What is the most common monthly connectivity plan?")
        print("4. Back to Main Menu")

        try:
            d = int(input("Enter your choice: "))
        except ValueError:
            print("Invalid input.")
            continue

        if d == 1:
            cat = upload["Daily Application Usage"].value_counts()
            print("Category Used the Most")
            print(cat)
            cat.plot(kind="bar", color="blue")
            plt.title("Category Most Used")
            plt.xlabel("categories")
            plt.tight_layout()
            plt.show()

        elif d == 2:
            prof = upload.groupby("Average Daily Screen-On Time (SOT)")["Digital Proficiency Self-Assessment"].value_counts().unstack()
            print("\nDigital Proficiency vs Screen Time")
            print(prof)
            prof.plot(kind="bar", stacked=True, color=["skyblue", "lightgreen"])
            plt.title("Proficiency vs Screen Time")
            plt.tight_layout()
            plt.show()
            print("\nUsers with higher digital proficiency tend to spend slightly more time on their phones, suggesting familiarity with technology leads to greater engagement. However, the correlation is weak, meaning proficiency alone is not the only factor driving screen time.")

        elif d == 3:
            conn = upload["Estimated Monthly Connectivity Overhead"].value_counts()
            print("\nEstimated Monthly Connectivity Overhead ")
            print(conn)
            conn.plot(kind="pie", autopct="%1.2f%%", colors=["pink", "skyblue", "lightgreen"])
            plt.title("Estimated Monthly Connectivity")
            plt.tight_layout()
            plt.show()

        elif d == 4:
            continue

        else:
            print("Invalid Choice. Please try again")

    elif op == 5:
        print("\nHEALTH AND WELLBEING IMPACT MENU")
        print("1. Does late-night phone usage affect circadian rhythm (sleep disruption)?")
        print("2. Which age group reports the highest perceived circadian impact?")
        print("3. Which users report the highest physical ergonomic strain?")
        print("4. Is there a link between meal-time phone use and high screen time?")
        print("5. Back to Main Menu")

        try:
            e = int(input("Enter Your Choice: "))
        except ValueError:
            print("Invalid input.")
            continue

        if e == 1:
            window_order = [
                'Morning (6:00 - 12:00)',
                'Afternoon (12:00 - 17:00)',
                'Evening (17:00 - 21:00)',
                'Late Night (21:00 - 2:00)',
            ]
            upload['Peak Activity Window'] = pd.Categorical(
                upload['Peak Activity Window'], categories=window_order, ordered=True)
            short_labels = ['Morning', 'Afternoon', 'Evening', 'Late Night']

            plt.figure(figsize=(10, 5))
            plt.suptitle('Does Late-Night Phone Usage Affect Circadian Rhythm?', fontsize=15, fontweight='bold')
            heatmap_data = pd.crosstab(upload['Peak Activity Window'], upload['Perceived Circadian Impact'])
            heatmap_data.index = short_labels
            plt.imshow(heatmap_data.values, cmap='RdYlBu_r', aspect='auto')
            plt.xticks(range(len(heatmap_data.columns)), [str(x) for x in heatmap_data.columns])
            plt.yticks(range(4), short_labels)
            plt.xlabel('Perceived Circadian Impact Score')
            plt.title('Peak Window vs\nSleep Disruption', fontweight='bold')
            plt.colorbar(label='Number of Users')
            for i in range(heatmap_data.shape[0]):
                for j in range(heatmap_data.shape[1]):
                    val = heatmap_data.values[i, j]
                    plt.text(j, i, str(val), ha='center', va='center',
                             fontweight='bold', color='white' if val > 8 else 'black')
            plt.tight_layout()
            plt.show()

        elif e == 2:
            avg = upload.groupby('Participant Age Group')['Perceived Circadian Impact'].mean().sort_values(ascending=False)
            plt.figure(figsize=(8, 5))
            bars = plt.bar(avg.index, avg.values, color='#E05C5C', edgecolor='white', width=0.5)
            plt.title('Average Circadian Impact by Age Group', fontweight='bold')
            plt.xlabel('Age Group')
            plt.ylabel('Avg Circadian Impact (1=Low, 5=High)')
            plt.ylim(0, 5)
            for bar, val in zip(bars, avg.values):
                plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.08,
                         f'{val:.2f}', ha='center', fontweight='bold')
            plt.tight_layout()
            plt.show()

        elif e == 3:
            avg = upload.groupby('Participant Age Group')['Physical Ergonomic Strain'].mean().sort_values()
            plt.figure(figsize=(8, 5))
            bars = plt.barh(avg.index, avg.values, color='#F0A500', edgecolor='white')
            plt.title('Average Ergonomic Strain by Age Group', fontweight='bold')
            plt.xlabel('Avg Ergonomic Strain (1=Low, 5=High)')
            plt.xlim(0, 5)
            for i, val in enumerate(avg.values):
                plt.text(val + 0.08, i, f'{val:.2f}', va='center', fontweight='bold')
            plt.tight_layout()
            plt.show()

        elif e == 4:
            sot_map = {
                'Tier 1: <2 hours'  : 1,
                'Tier 2: 2-4 hours' : 2,
                'Tier 3: 4-6 hours' : 3,
                'Tier 4: 6-8 hours' : 4,
                'Tier 5: 8+ hours'  : 5,
            }
            upload['SOT_Numeric'] = upload['Average Daily Screen-On Time (SOT)'].map(sot_map)
            x = upload['Meal-Time Device Interaction']
            y = upload['SOT_Numeric']
            np.random.seed(42)
            jitter = np.random.uniform(-0.1, 0.1, size=len(upload))
            plt.figure(figsize=(8, 5))
            plt.scatter(x + jitter, y + jitter, color='#5C9BE0',
                        alpha=0.5, edgecolors='white', s=70)
            z = np.polyfit(x, y, 1)
            p = np.poly1d(z)
            x_line = np.linspace(1, 5, 100)
            plt.plot(x_line, p(x_line), color='#E05C5C', linewidth=2,
                     linestyle='--', label='Trend line')
            plt.xticks([1, 2, 3, 4, 5])
            plt.yticks([1, 2, 3, 4, 5], ['<2 hrs', '2-4 hrs', '4-6 hrs', '6-8 hrs', '8+ hrs'])
            plt.xlabel('Meal-Time Device Interaction (1=Low, 5=High)')
            plt.ylabel('Screen-On Time')
            plt.title('Meal-Time Phone Use vs Screen Time', fontweight='bold')
            plt.legend()
            plt.tight_layout()
            plt.savefig('q4_meal_sot.png', dpi=150)
            plt.show()
            print("Chart saved as q4_meal_sot.png")

        elif e == 5:
            continue

        else:
            print("Invalid choice! Please Enter a Valid option")

    elif op == 6:
        print("Thanku! Bye Bye")
        break

    else:
        print("Invalid Choice! Please Enter a Valid option")
```
